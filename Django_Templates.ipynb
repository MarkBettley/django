# Django Templates

En este ejercicio se utilizaron filtros, ciclos for, herencia de templates
e include con variables.

Se creó un template base reutilizable y templates específicos para mostrar
los productos del eCommerce.


## Template base reutilizable

```python
<!DOCTYPE html>
<html lang="es">
<head>
    <meta charset="UTF-8">
    <title>
        {% block title %}
            Mi eCommerce
        {% endblock %}
    </title>
</head>
<body>

    <header>
        <h1>Mi eCommerce</h1>

        <nav>
            <a href="{% url 'product-list' %}">
                Productos
            </a>

            <a href="{% url 'my-products' %}">
                Mis productos
            </a>

            <a href="{% url 'register' %}">
                Registrarse
            </a>
        </nav>

        <hr>
    </header>

    <main>
        {% block content %}
        {% endblock %}
    </main>

    <hr>

    <footer>
        <p>Proyecto Django</p>
    </footer>

</body>
</html>
```

## Template reutilizable con filtros

```python
<article>

    <h2>
        {{ product.name|upper }}
    </h2>

    <p>
        <strong>Precio:</strong>
        ${{ product.price }}
    </p>

    <p>
        <strong>Descripción:</strong>
        {{ product.description|truncatechars:40 }}
    </p>

    <p>
        <strong>Vendedor:</strong>
        {{ product.seller|title }}
    </p>

    <p>
        <strong>Color:</strong>
        {{ product.color|lower }}
    </p>

    <p>
        <strong>Dimensiones:</strong>
        {{ product.product_dimensions }}
    </p>

    <a href="{% url 'product-detail' product.pk %}">
        Ver detalle
    </a>

</article>
```

## Ciclo for, divisibleby e include

```python
{% extends "ejercicios/base.html" %}

{% block title %}
    Lista de productos
{% endblock %}

{% block content %}

    <h1>Lista de productos</h1>

    <a href="{% url 'product-create' %}">
        Crear nuevo producto
    </a>

    <hr>

    {% if products %}

        {% for product in products %}

            <p>
                Producto número:
                {{ forloop.counter }}
            </p>

            {% if forloop.counter|divisibleby:"5" %}
                <p>
                    Esta iteración es divisible entre 5.
                </p>
            {% endif %}

            {% include "ejercicios/product_card.html" with product=product %}

            <hr>

        {% endfor %}

    {% else %}

        <p>
            No hay productos registrados.
        </p>

    {% endif %}

{% endblock %}
```

## Herencia en el detalle del producto

```python
{% extends "ejercicios/base.html" %}

{% block title %}
    Detalle de {{ product.name }}
{% endblock %}

{% block content %}

    <h1>
        {{ product.name|upper }}
    </h1>

    <p>
        <strong>Precio:</strong>
        ${{ product.price }}
    </p>

    <p>
        <strong>Descripción:</strong>
        {{ product.description }}
    </p>

    <p>
        <strong>Vendedor:</strong>
        {{ product.seller|title }}
    </p>

    <p>
        <strong>Color:</strong>
        {{ product.color|lower }}
    </p>

    <p>
        <strong>Dimensiones:</strong>
        {{ product.product_dimensions }}
    </p>

    <hr>

    <a href="{% url 'product-update' product.pk %}">
        Editar producto
    </a>

    <a href="{% url 'product-delete' product.pk %}">
        Eliminar producto
    </a>

    <a href="{% url 'product-list' %}">
        Volver a la lista
    </a>

{% endblock %}
```

## Pruebas realizadas

Se comprobó desde el navegador que:

1. base.html se hereda correctamente en product_list.html y product_detail.html.
2. El ciclo for recorre correctamente los productos.
3. forloop.counter muestra el número de cada iteración.
4. El filtro divisibleby detecta cada quinta iteración.
5. En el producto número 5 aparece:
   "Esta iteración es divisible entre 5."
6. product_card.html se incluye mediante include pasando la variable product.
7. Los filtros upper, truncatechars, title y lower funcionan correctamente.
8. El código común de navegación y estructura se reutiliza mediante base.html.

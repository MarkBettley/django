# Blockchain con Python

## Ejercicio

El objetivo de esta actividad es crear una blockchain con 5 bloques de transacciones utilizando Python.

Se utilizan las clases `Wallet`, `Transaction`, `Block` y `Blockchain`.

## 1. Clases utilizadas

Las clases permiten crear wallets, realizar transacciones, almacenarlas en bloques y enlazar los bloques mediante hashes SHA-256.

```python
from blockchain_clases import (
    Wallet,
    Transaction,
    Block,
    Blockchain,
)
```

## 2. Creación de wallets

Se crean cuatro wallets con diferentes saldos iniciales.

```python
wallet_marco = Wallet('Marco', 1000)
wallet_ana = Wallet('Ana', 800)
wallet_carlos = Wallet('Carlos', 600)
wallet_laura = Wallet('Laura', 500)

print(wallet_marco)
print(wallet_ana)
print(wallet_carlos)
print(wallet_laura)
```

## 3. Blockchain y transacciones

Se crea una instancia de `Blockchain` y se realizan transacciones entre las wallets. Las transacciones se almacenan en cinco bloques.

```python
blockchain = Blockchain()

# Bloque 1
t1 = wallet_marco.crear_transaccion(wallet_ana, 100)
t2 = wallet_ana.crear_transaccion(wallet_carlos, 50)
bloque_1 = Block(1, [t1, t2])
blockchain.agregar_bloque(bloque_1)

# Bloque 2
t3 = wallet_carlos.crear_transaccion(wallet_laura, 75)
t4 = wallet_marco.crear_transaccion(wallet_carlos, 120)
bloque_2 = Block(2, [t3, t4])
blockchain.agregar_bloque(bloque_2)

# Bloque 3
t5 = wallet_laura.crear_transaccion(wallet_ana, 40)
t6 = wallet_carlos.crear_transaccion(wallet_marco, 60)
bloque_3 = Block(3, [t5, t6])
blockchain.agregar_bloque(bloque_3)

# Bloque 4
t7 = wallet_ana.crear_transaccion(wallet_laura, 90)
t8 = wallet_marco.crear_transaccion(wallet_laura, 30)
bloque_4 = Block(4, [t7, t8])
blockchain.agregar_bloque(bloque_4)

# Bloque 5
t9 = wallet_laura.crear_transaccion(wallet_carlos, 110)
t10 = wallet_carlos.crear_transaccion(wallet_ana, 25)
bloque_5 = Block(5, [t9, t10])
blockchain.agregar_bloque(bloque_5)
```

## 4. Mostrar la blockchain

Cada bloque muestra sus transacciones, su hash y el hash del bloque anterior.

```python
blockchain.mostrar()
```

## 5. Verificación

Finalmente se comprueba que la blockchain contiene exactamente cinco bloques y que la cadena conserva su integridad.

```python
print('Cantidad de bloques:', len(blockchain.bloques))
print('Blockchain válida:', blockchain.es_valida())

assert len(blockchain.bloques) == 5
assert blockchain.es_valida() is True
```

# Django Models & Admin

En este ejercicio se crearon 500 productos de prueba de forma automática
utilizando un ciclo y bulk_create().

Después se creó un fixture, se eliminaron los productos de la base de datos
y finalmente se restauraron utilizando loaddata.


## Código de creación masiva

```python
from .models import ProductModel


def create_products():
    products = []

    for i in range(1, 501):
        product = ProductModel(
            name=f"Producto {i}",
            price=100 + i,
            description=f"Descripción del producto de prueba {i}",
            seller=f"Vendedor {(i % 10) + 1}",
            color=f"Color {(i % 5) + 1}",
            product_dimensions=f"{10 + (i % 20)} x {20 + (i % 20)} x 5 cm",
        )

        products.append(product)

    ProductModel.objects.bulk_create(products)

    print(f"Productos creados: {ProductModel.objects.count()}")
```

## Fixture

Se creó el archivo:

ejercicios/fixtures/products/500Products.json

El fixture contiene exactamente 500 productos.


## Pruebas realizadas

1. Base de datos inicial: 0 productos.
2. bulk_create: 500 productos creados.
3. Fixture generado: 500 productos.
4. Se eliminaron todos los productos.
5. Base de datos después de eliminar: 0 productos.
6. Se cargó el fixture con loaddata.
7. Resultado final: 500 productos restaurados.


## Comandos utilizados

Crear fixture:

python manage.py dumpdata ejercicios.ProductModel --indent 4 --format json > ejercicios/fixtures/products/500Products.json

Cargar fixture:

python manage.py loaddata ejercicios/fixtures/products/500Products.json

## Clases y objetos

En este ejercicio se utilizan las clases `Producto` y `ProductoInternacional`.
La segunda clase hereda los atributos y métodos de la clase `Producto`.


## Definición de las clases

```python
class Producto:
    def __init__(self, nombre, precio, descripcion):
        self.nombre = nombre
        self.precio = precio
        self.descripcion = descripcion

    def calcular_total(self, cantidad):
        return cantidad * self.precio


class ProductoInternacional(Producto):
    def __init__(self, nombre, precio, descripcion, tasa_internacional):
        super().__init__(nombre, precio, descripcion)
        self.tasa_internacional = tasa_internacional

    def calcular_total_con_tasa(self, cantidad):
        return self.calcular_total(cantidad) * (1 + self.tasa_internacional)
```

## Instanciación y uso de los objetos

```python
# Ejemplos de instanciación de objetos

producto = Producto(
    "Camiseta",
    50.0,
    "Camiseta de algodón"
)

producto_internacional = ProductoInternacional(
    "Tenis",
    100.0,
    "Tenis deportivos importados",
    0.16
)

print("Producto nacional:")
print("Nombre:", producto.nombre)
print("Precio:", producto.precio)
print("Descripción:", producto.descripcion)
print("Total por 10 unidades:", producto.calcular_total(10))

print("\nProducto internacional:")
print("Nombre:", producto_internacional.nombre)
print("Precio:", producto_internacional.precio)
print("Descripción:", producto_internacional.descripcion)
print("Tasa internacional:", producto_internacional.tasa_internacional)
print("Total con tasa por 10 unidades:", producto_internacional.calcular_total_con_tasa(10))
```

# Visualización de Datos

## Dashboard Financiero para un eCommerce

El objetivo de esta actividad es crear un dashboard financiero utilizando datos de ventas de un eCommerce, Plotly y Dash.

El dashboard permite analizar ingresos, costos, utilidad, margen, ventas por categoría y productos vendidos.

## 1. Diseño del dashboard

Se planteó un dashboard compuesto por:

- Indicador de ingresos.
- Indicador de costos.
- Indicador de utilidad.
- Indicador de margen.
- Evolución de ingresos y utilidad.
- Ingresos por categoría.
- Top de productos vendidos.
- Comparación de ingresos y utilidad por categoría.
- Filtro por rango de fechas.
- Filtro por categoría.

El boceto completo se encuentra en `boceto_dashboard.txt`.

## 2. Importar librerías y datos

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv(
    'ventas_ecommerce.csv',
    parse_dates=['fecha'],
)

df.head()
```

## 3. Indicadores financieros

Se calculan los ingresos, costos, utilidad y margen de utilidad del eCommerce.

```python
ingresos = df['ingresos'].sum()
costos = df['costo'].sum()
utilidad = df['utilidad'].sum()
margen = utilidad / ingresos * 100

print(f'Ingresos: ${ingresos:,.2f}')
print(f'Costos: ${costos:,.2f}')
print(f'Utilidad: ${utilidad:,.2f}')
print(f'Margen: {margen:.2f}%')
```

## 4. Evolución de ingresos y utilidad

```python
ventas_tiempo = (
    df.groupby('fecha', as_index=False)
    .agg(
        ingresos=('ingresos', 'sum'),
        utilidad=('utilidad', 'sum'),
    )
)

figura_tiempo = px.line(
    ventas_tiempo,
    x='fecha',
    y=['ingresos', 'utilidad'],
    title='Evolución de ingresos y utilidad',
)

figura_tiempo.show()
```

## 5. Ingresos por categoría

```python
ventas_categoria = (
    df.groupby('categoria', as_index=False)['ingresos']
    .sum()
    .sort_values('ingresos', ascending=False)
)

figura_categorias = px.bar(
    ventas_categoria,
    x='categoria',
    y='ingresos',
    title='Ingresos por categoría',
)

figura_categorias.show()
```

## 6. Productos más vendidos

```python
productos = (
    df.groupby('producto', as_index=False)['unidades']
    .sum()
    .sort_values('unidades', ascending=False)
    .head(10)
)

figura_productos = px.bar(
    productos,
    x='unidades',
    y='producto',
    orientation='h',
    title='Top 10 productos por unidades vendidas',
)

figura_productos.show()
```

## 7. Ingresos vs utilidad por categoría

```python
comparacion = (
    df.groupby('categoria', as_index=False)
    .agg(
        ingresos=('ingresos', 'sum'),
        utilidad=('utilidad', 'sum'),
    )
)

figura_comparacion = px.scatter(
    comparacion,
    x='ingresos',
    y='utilidad',
    size='ingresos',
    text='categoria',
    title='Ingresos vs utilidad por categoría',
)

figura_comparacion.show()
```

## 8. Interactividad con Dash

El archivo `dashboard_financiero.py` implementa el dashboard completo mediante Dash.

El usuario puede seleccionar un rango de fechas y una o varias categorías. Cada selección actualiza automáticamente los cuatro indicadores financieros y las cuatro gráficas.

## Conclusión

El dashboard permite visualizar de forma interactiva el desempeño financiero del eCommerce. Plotly se utiliza para generar las visualizaciones y Dash para integrar las gráficas, indicadores y filtros en una aplicación web.
